Data cleaning

```python
titanic_df.head()
```

```python
plt.figure(figsize=(12,8))
sns.boxplot(x='class',y='age',data=titanic_df)
# here 50% quantile(or 2nd quartile) is a median 
```

```python
def impute_age(cols):
    age=cols[0]
    pclass=cols[1]

    if pd.isnull(age):
        if(pclass==1):
            return 37
        elif(pclass==2):
            return 29
        else:
            return 24
    else:
        return age
```

```python
titanic_df['age']= titanic_df[['age','pclass']].apply(impute_age,axis=1)
```

```python
sns.heatmap(titanic_df.isnull())
```

```python
titanic_df.isnull().sum()
```

```python
titanic_df.head()
```

```python
sns.countplot(x='embarked',data=titanic_df)
```

```python
titanic_df.fillna({'embarked':'S'},inplace=True)
titanic_df.fillna({'embark_town':'Sowthampton'},inplace=True)
```

```python
titanic_df.isnull().sum()
```

```python
titanic_df.drop('deck',axis=1,inplace=True)
```

```python
sns.heatmap(titanic_df.isnull())
```

{converting categorical features}
we will need to convert categorical features to dummy variables using pandas otherwise machine learning algorithm wont be able to directly take in those feature as inputs

```python
titanic_df.head()
```

```python
embark = pd.get_dummies(titanic_df['embarked'],drop_first=True,dtype=int).head()
embark
```

```python
sex = pd.get_dummies(titanic_df['sex'],drop_first=True,dtype=int).head()
sex
```

```python
titanic_df.drop(['sex','embarked','who','class','adult_male','embark_town','alive','alone'],axis=1,inplace=True)
```

```python
titanic_df.columns
```

```python
titanic_df=pd.concat([titanic_df,sex,embark],axis=1)
```
